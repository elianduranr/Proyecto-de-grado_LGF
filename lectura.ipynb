Una misma semana de producción puede aparecer en varios registros porque se pronosticó desde diferentes orígenes y horizontes. Por eso **no se debe sumar `y` de toda la tabla como si fueran producciones distintas**.

### Teoría previa: versiones anteriores, no horizontes anteriores

**Hay dos preguntas distintas:** el horizonte dice **para qué semana** pronosticamos; `teoria_previa_1` y `teoria_previa_2` dicen **qué proyectaban versiones anteriores del plano para esa misma semana**. El 1 y el 2 cuentan versiones hacia atrás, no semanas objetivo H1 y H2.

Para el registro de DINO, mantenemos fija la semana del **31 de agosto al 6 de septiembre de 2026**. Al emitir el pronóstico el **3 de agosto**, el proceso seleccionó el plano de la semana 29. Al mirar las dos versiones anteriores, encontramos lo siguiente:

| Papel en esta fila | Plano | Disponibilidad usada por el proceso | Teoría calculada para el 31 de agosto–6 de septiembre | Grupos sin cobertura | Valor que se habilita |
|---|---|---|---:|---:|---|
| Teoría del plano seleccionado | Semana 29 | 28 de julio | 27.916,61 tallos | 2 | `teoria` conserva la suma parcial; `teoria_modelo` queda vacío. |
| Una versión atrás | Semana 28 | 21 de julio | 29.979,80 tallos | 2 | `teoria_previa_1` queda vacío. |
| Dos versiones atrás | Semana 27 | 14 de julio | 31.268,38 tallos | 2 | `teoria_previa_2` queda vacío. |

**No confundir los dos alcances:** el modelo semanal evalúa H1–H5 desde su fecha de origen, pero las proyecciones teóricas guardadas abarcan **10 semanas por plano** (`HORIZONTE = 10`). Además, el generador teórico cuenta desde la semana siguiente a `fecha_plano`, mientras el modelo cuenta H1 como la propia semana de `origen`. Por eso sus números de horizonte no se pueden intercambiar directamente.

En este ejemplo, el 31 de agosto es **H5 para el modelo emitido el 3 de agosto**, pero es el horizonte teórico **7 del plano del 13 de julio**, **8 del plano del 6 de julio** y **9 del plano del 29 de junio**. Los tres existen dentro de las diez semanas generadas. Las teorías previas están vacías por cobertura incompleta, no porque se haya superado un límite de cinco semanas. Si un plano anterior no alcanzara la semana objetivo, tampoco habría una teoría previa disponible para esa semana.

**Aquí sí existían proyecciones anteriores.** Las columnas previas están vacías porque el código conserva únicamente valores anteriores con cobertura completa. En cambio, `teoria` conserva también la suma parcial del plano seleccionado. Esa diferencia explica por qué podemos ver un número en `teoria` y vacíos en las dos columnas previas.

En otros registros, una teoría previa también puede faltar porque no exista esa versión anterior para la serie y semana objetivo. El cálculo toma las versiones anteriores en orden de disponibilidad; **no salta una versión incompleta para buscar otra completa más antigua**. Tampoco tienen que estar separadas exactamente por una semana.

**Ejemplo inventado con cobertura completa:** si el plano seleccionado proyecta 28.000 tallos, la versión anterior proyectaba 30.000 y la anterior a esa 31.000, entonces `teoria_modelo = 28.000`, `teoria_previa_1 = 30.000`, `teoria_previa_2 = 31.000` y `revision_teoria = 28.000 − 30.000 = −2.000 tallos`. Significa que la proyección para **la misma semana** se revisó a la baja en 2.000 tallos. No significa que la producción real haya caído esa cantidad.

En el registro real de DINO, `revision_teoria` y `desviacion_reciente_teoria` están vacíos porque requieren teoría habilitada. La fila sigue siendo admisible por su historial de producción; eso no convierte la suma parcial en una teoría completa.

### Seno y coseno: cómo se calculan y qué significan

**Estas dos columnas son una forma de escribir la fecha para que el modelo entienda en qué época del año estamos. No calculan producción.** Se obtienen de la **semana objetivo**, no de la semana de origen ni del horizonte H1–H5.

#### 1. Primero sacamos el número de semana del año

En el ejemplo de DINO queremos pronosticar la semana que empieza el **31 de agosto de 2026**. Esa fecha pertenece a la **semana 36 del año**, según el calendario ISO (semanas que empiezan el lunes).

Aquí tenemos dos números distintos: **H5** significa que estamos pronosticando la quinta semana contando desde el origen; **36** es la posición de la semana objetivo dentro del año. Para calcular seno y coseno utilizamos **36**, no 5.

#### 2. Convertimos esa semana en una posición alrededor de un círculo

Piensa en un reloj: después de dar una vuelta, volvemos al comienzo. El calendario también vuelve a empezar después de diciembre.

Dividimos el círculo en **52 pasos**, uno por semana. Para ubicar la semana 36:

- Dividimos **36 / 52 = 0,6923**. Eso equivale al **69,23 % de una vuelta** en esta representación del calendario; no es un porcentaje de producción ni una medida exacta de los días transcurridos del año.
- Una vuelta tiene **360 grados**, así que hacemos **0,6923 × 360 = 249,23 grados**.
- Imagina una aguja que sale del centro: empieza apuntando hacia arriba y gira en sentido horario hasta esa posición. Para la semana 36 termina apuntando hacia abajo y a la izquierda.

#### 3. Seno y coseno describen dónde termina la aguja

Si la aguja mide una unidad, podemos describir su punta con dos números:

- **`semana_seno`: posición horizontal.** Derecha es positiva; izquierda es negativa.
- **`semana_coseno`: posición vertical.** Arriba es positiva; abajo es negativa.

Las funciones matemáticas *seno* y *coseno* hacen esa conversión del ángulo a las dos posiciones. Python las calcula automáticamente; no hay que inventarlas ni estimarlas con datos de producción.

```text
                           ARRIBA
                         semana 52
                       seno=0, cos=1
                              |
     IZQUIERDA ---------------+--------------- DERECHA
     semana 39                |                semana 13
     seno=-1, cos=0           |                seno=1, cos=0
                            ABAJO
                          semana 26
                        seno=0, cos=-1
```

**Para la semana 36**, los resultados son:

| Columna | Cálculo con el ángulo en grados | Resultado aproximado | Significado |
|---|---|---:|---|
| `semana_seno` | seno de 249,23° | −0,935 | La punta está hacia la izquierda del centro. |
| `semana_coseno` | coseno de 249,23° | −0,355 | La punta está un poco por debajo del centro. |

**No leas −0,935 como −93,5 % de producción.** Ambos números solo indican posición en el calendario. Tampoco significan que la producción vaya a bajar.

#### 4. Así se hace en Python

Python utiliza **radianes** para los ángulos: es otra unidad, igual que medir una distancia en metros o centímetros. **360 grados equivalen a 2 × π radianes**, con π aproximadamente igual a 3,1416. Por eso el cálculo del notebook es:

```python
import numpy as np
import pandas as pd

fecha_objetivo = pd.Timestamp("2026-08-31")
semana_del_anio = fecha_objetivo.isocalendar().week  # 36, no H5
angulo = 2 * np.pi * semana_del_anio / 52           # 4,3499 radianes = 249,23°

semana_seno = np.sin(angulo)      # aproximadamente -0,935
semana_coseno = np.cos(angulo)    # aproximadamente -0,355
```

Este bloque ilustra el cálculo; esas columnas ya vienen calculadas en el parquet.

#### 5. ¿Por qué usamos los dos números y para qué sirven?

**Para conservar la cercanía entre diciembre y enero.** Si usamos únicamente el número de semana, 52 y 1 parecen estar muy lejos. En el círculo, la semana 52 tiene valores **(0; 1)** y la semana 1 tiene **(0,121; 0,993)**: quedan cerca, como sucede en el calendario.

**Usamos ambos porque uno solo puede ser ambiguo.** Por ejemplo, las semanas 13 y 39 tienen el mismo coseno (0), pero están en épocas diferentes. El seno permite distinguirlas: en una vale 1 y en la otra −1.

El modelo recibe esos dos números junto con los demás insumos y puede aprender si hay diferencias recurrentes de producción según la época del año. **Estas columnas no dicen por sí solas cuánto se producirá ni garantizan una mejora del modelo.** Solo aportan información del calendario, sin consultar clima ni producción futura.

*Detalle del cálculo actual:* se utiliza un ciclo fijo de 52 semanas. Algunos años ISO tienen semana 53; con esta fórmula recibe la misma posición que la semana 1. Es una aproximación del calendario anual.

**Cómo interpretar los vacíos:** `NaN`, `NaT` o `<NA>` significan dato ausente o no utilizable, no cero. Por ejemplo, que falte una teoría previa no significa que esa versión pronosticara cero tallos.

**Alcance de esta tabla:** es la base semanal común. Los notebooks posteriores añaden variables específicas, como tendencias de corte diario. Que una columna esté aquí no significa que todos los modelos la utilicen. Tampoco están aquí el pronóstico final del modelo ni el estimado del ingeniero: para comparar errores hay que consultar las salidas de predicción correspondientes. La división entre entrenamiento y prueba se aplica después, por fechas.
